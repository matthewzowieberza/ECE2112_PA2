### **A. REPRODUCIBLE NORMALIZATION PROBLEM**

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
xmean = np.mean(X)
xmean
```

```python
sd = np.std(X)
sd
```

```python
X_normalized = (X - xmean) / sd
X_normalized
```

```python
xnormm = np.mean(X_normalized)
xnormm
```

```python
xnormstd = np.std(X_normalized)
xnormstd
```

```python
print ("Original Array: ")
print (X)
print ("\nMean: ")
print (xmean)
print ("\nStandard Deviation")
print (sd)
print ("\n-----------------------------------------------------------------")
print ("\nNormalized Array: ")
print (X_normalized)
print ("\nNormalized Mean:")
print (xnormm)
print ("\nNormalized Standard Deviation:")
print (xnormstd)
```

```python
np.save("X_normalized.npy", X_normalized)
```

## **B. CUBES DIVISIBLE BY 4 PROBLEM**

```python
c = np.arange(1,101)
c
```

```python
C = (c**3)
C
```

```python
Cc = C.reshape(10, 10)
Cc
```

```python
div_by_4 = C[C % 4 == 0]
div_by_4
```

```python
print ("Shape of C: ")
print (Cc.shape)
print ("\nArray Divisible by 4: ")
print (div_by_4)
print ("\nArray Size: ")
print (div_by_4.size)
```

```python
np.save("div_by_4.npy", div_by_4)
```

### **C. ABOVE-MEAN SQUARES PROBLEM**

```python
s = np.arange(1,37)
s
```

```python
s = s.reshape(6,6)
s
```

```python
S = s**2
S
```

```python
S_mean = np.mean(S)
S_mean
```

```python
above_mean = S[S > S_mean]
above_mean
```

```python
print ("Array S: ")
print (S)
print ("\nArray Mean: ")
print (S_mean)
print ("Above Mean: ")
print (above_mean)
print ("Number of Elements: ")
print (above_mean.size)
```

```python
np.save("above_mean.npy", above_mean)
```
